```python
# 变量的类型

a1=1
a2=1.0
a3="1"
a4=True
a5=None
a6=[1,2,3]
a7=(1,2,3)
a8={1,2,3}
a9={"a":1,"b":2,"c":3}
print(type(a1),type(a2),type(a3),type(a4),type(a5),type(a6),type(a7),type(a8),type(a9))
```

```python
# 运算符

## 算术运算符

x=10
y=3

print(x+y,x-y,x*y,x/y,x//y,x%y,x**y)  # 加法


## 比较运算符
print(x>y,x<y,x>=y,x<=y,x==y,x!=y)  


## 逻辑运算符
print(x>5 and y<5,x>5 or y<5,not(x>5 and y<5)) 

## 成员运算符

print(5 in [1,2,3,4,5], 6 not in [1,2,3,4,5])
```

```python
# 流程控制

## 顺序结构

print("这是第一句话")
print("这是第二句话")
print("这是第三句话")
```

```python
## 分支结构

age=20
if age < 18:
    print("未成年")
else:
    print("成年")
```

```python
## 循环结构

for i in range(5):
    print(i)


体力值=100

while 体力值>0:
    print("体力值为：", 体力值, "继续战斗")
    体力值-=10
```

```python
# 类和对象

class 大学生：
    def __init__(self, 姓名, 年龄，性别，学校="某大学" ，专业="某专业"):
        self.姓名 = 姓名
        self.年龄 = 年龄
        self.性别 = 性别
        self.学校 = 学校
        self.专业 = 专业

    def 打招呼（self):
        print("大家好，我是", self.姓名, "，今年", self.年龄, "岁。")

    def 学习（self, 科目):
        print(self.姓名, "正在学习", 科目, "。")
```

```python
李逍遥 = 大学生("李逍遥", 20, "男"，"大理大学", "武术")
李逍遥.打招呼()
```
